## Era 2: Jupyter Notebooks (2014-present)

### The game-changer
- **Born 2014** when Project Jupyter spun off from IPython
- By **2018**, 2.5 million notebooks on GitHub; *Nature* called it a cornerstone of scientific computing
- **Interactive:** Run code cell-by-cell, see results immediately
- **Inline plots:** No more saving PNGs and opening them in a viewer
- **Narrative + code:** Mix markdown explanations with executable code
- **Exploration:** Try something, see the output, adjust — all in one place

### What changed
- **Pandas arrived** in the data science mainstream
- `pd.read_csv()` replaced 15 lines of csv module code
- `df.groupby()` replaced 30 lines of dictionary aggregation
- `%matplotlib inline` meant instant visual feedback

*Open `era_2_jupyter.ipynb` to see the interactive version.*

````python
# The same data loading — ONE LINE instead of 15
import matplotlib.pyplot as plt
%matplotlib inline
plt.style.use('ggplot')

df = pd.read_csv('data/marketing_campaign.csv', parse_dates=['date'])
df.head()
````

````python
# The same aggregation — 5 lines instead of 30
channel_summary = df.groupby('channel').agg(
    spend=('spend', 'sum'),
    revenue=('revenue', 'sum'),
    clicks=('clicks', 'sum'),
    impressions=('impressions', 'sum'),
)
channel_summary['ROI (%)'] = ((channel_summary['revenue'] - channel_summary['spend']) / channel_summary['spend'] * 100).round(1)
channel_summary
````

````python
# Inline plot — see it right here, iterate instantly
channel_summary[['spend', 'revenue']].plot(kind='bar', color=['#cc4444', '#44aa44'], figsize=(10, 5))
plt.title('Spend vs Revenue by Channel')
plt.ylabel('Amount ($)')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
````

### Era 2 Impact

| Metric | Era 1 | Era 2 |
|:---|:---:|:---:|
| Lines of code | ~160 | ~50 |
| Time to first chart | 30+ min | 5 min |
| Feedback loop | Save → terminal → image viewer | Instant, inline |
| Data exploration | Painful | Natural |

**Key insight:** Jupyter didn't just make coding faster — it changed *how* data scientists think. Instead of planning everything upfront, you explore iteratively.

## Era 3: The IDE Revolution (2016-present)

### VS Code & PyCharm become the standard

## Era 4: Stack Overflow (2008-2023)

### The copy-paste workflow

Stack Overflow launched in **2008**, peaked at **200K questions/month in 2014**, and was THE developer resource until AI tools arrived. Traffic dropped **78% between 2024-2025**.

The real skill became *searching*, not *writing*:

1. Google: "matplotlib two y-axes on same plot"
2. Open the Stack Overflow result (450K views)
3. Scroll past the question to the green-checkmark answer
4. Copy the code block
5. Paste into your script, change variable names
6. Run it, see if it works
7. If not, try the second-highest answer

### The good and the bad
- **Good:** Access to millions of solved problems
- **Bad:** No context — is this answer from 2014 still the best approach?
- **Bad:** Frankenstein code — each block from a different source, different style
- **Ugly:** Spending 45 min finding the right answer for a 2-line fix

````python
# Look at the SO-style comments in Era 4's code
era4_code = open('era_4_stackoverflow.py').read()
so_refs = [line.strip() for line in era4_code.splitlines() if 'SO:' in line]
print(f"Stack Overflow references in the code: {len(so_refs)}")
print()
for ref in so_refs:
    print(f"  {ref}")
````

## Era 5: ChatGPT (Late 2022-present)

### The paradigm shift

ChatGPT launched **November 30, 2022** and changed everything overnight.

Instead of searching for answers, you **describe what you want:**

```
"I have a CSV file called marketing_campaign.csv with columns:
 date, channel, campaign_name, impressions, clicks, conversions,
 spend, revenue.

 Can you write a Python script that:
 1. Loads the data and shows basic statistics
 2. Creates a bar chart of spend vs revenue by channel  
 3. Creates a monthly revenue trend line chart
 4. Calculates ROI and CTR for each channel
 Use pandas and matplotlib. Make the plots look professional."
```

**Response time:** ~15 seconds

**Result:** A complete, working script. On the first try.

### What changed
- **From HOW to WHAT:** You describe the outcome, AI writes the implementation
- **No more searching:** No SO tabs, no documentation deep-dives
- **Consistent style:** The generated code follows one consistent pattern
- **But:** You still copy-paste from a browser into your editor

## Era 6: Cursor / AI IDE (2023-present)

### AI moves *into* your editor

Founded in **2022**, Cursor hit **$100M ARR by January 2025** and became the fastest-growing AI coding tool.

Key differences from ChatGPT:

- **Context-aware:** The AI sees your *entire codebase* — all files, all imports, your data
- **Inline editing:** Select code, press `Cmd+K`, describe the change → it modifies in-place
- **Tab completion:** AI predicts the next several lines based on context
- **No copy-paste:** Everything happens inside the editor
- **Multi-file awareness:** "Add error handling to all API endpoints" → changes 5 files at once

### The workflow
1. Write a comment: `# Create a 4-panel dashboard comparing channel metrics`
2. Press Tab → entire function appears
3. If it's not quite right: select, `Cmd+K`, "make the colors match our brand palette"
4. The AI *already knows* your PALETTE dict from line 20

### Result
More sophisticated analysis because the *cost of trying things* dropped to near zero.

## The Big Picture

### What changed

```
Era 1 (~2010): You write every character         → "I am the coder"
Era 2 (2014):  You explore interactively          → "I think in small steps"
Era 3 (2016):  Your editor helps you write        → "I code with guardrails"
Era 4 (2008):  You find and adapt solutions       → "I am a great searcher"
Era 5 (2022):  You describe, AI writes            → "I am the specifier"
Era 6 (2023):  AI is your pair programmer         → "I am the reviewer"
Era 7 (2025):  AI builds, you architect           → "I am the thinker"
```

### What stayed the same

Across all 7 eras, you still need to:

1. **Understand the problem** — What question are we trying to answer?
2. **Know the data** — What does each column mean? What are the edge cases?
3. **Think critically** — Does this result make sense? Is the chart telling the truth?
4. **Communicate findings** — The chart is not the deliverable. The *insight* is.

### The skill shift

| Old Skills (still useful) | New Skills (increasingly important) |
|:---|:---|
| Syntax memorization | Problem decomposition |
| Library API knowledge | Prompt engineering |
| Debugging line-by-line | Code review & verification |
| Writing code from scratch | Describing what you want clearly |
| Searching Stack Overflow | Evaluating AI-generated solutions |

## Summary: Lines of Code Written by Human

A rough comparison for the *same analysis task*:

````python
import matplotlib.pyplot as plt
import numpy as np

eras = ['Era 1\nNotepad\n(~2010)', 'Era 2\nJupyter\n(2014)', 'Era 3\nIDE\n(2016)', 
        'Era 4\nStack OF\n(2008)', 'Era 5\nChatGPT\n(2022)', 'Era 6\nCursor\n(2023)',
        'Era 7\nClaude Code\n(2025)']
human_lines = [160, 50, 95, 30, 5, 10, 0]
ai_lines = [0, 0, 0, 0, 100, 150, 200]
so_lines = [0, 0, 0, 90, 0, 0, 0]

fig, ax = plt.subplots(figsize=(14, 7))

x = np.arange(len(eras))
width = 0.55

bars1 = ax.bar(x, human_lines, width, label='Written by human', color='#4C6EF5', alpha=0.85)
bars2 = ax.bar(x, so_lines, width, bottom=human_lines, label='Copied from Stack Overflow', color='#FFA94D', alpha=0.85)
bars3 = ax.bar(x, ai_lines, width, bottom=[h+s for h, s in zip(human_lines, so_lines)],
               label='Generated by AI', color='#51CF66', alpha=0.85)

ax.set_ylabel('Lines of Code', fontsize=13)
ax.set_title('Who Writes the Code? The Shift from Human to AI', fontsize=15, fontweight='bold')
ax.set_xticks(x)
ax.set_xticklabels(eras, fontsize=10)
ax.legend(fontsize=11, loc='upper left')
ax.grid(axis='y', alpha=0.3)
ax.set_ylim(0, 250)

plt.tight_layout()
plt.savefig('plots/summary_who_writes_code.png', dpi=150, bbox_inches='tight')
plt.show()
````
